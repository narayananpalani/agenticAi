````python
# Week 1 - Hello LLM / Lab Step 2
#
# This version uses Ollama instead of the local Ollama service.
# The model runs locally, so there is no API key or API credit requirement.
# Frequently Asked Interview Questions:
# Q: How do you measure the computational cost or token generation of a local inference request?
# A: By extracting the `prompt_eval_count` (input tokens) and `eval_count` (completion tokens) attributes directly from the response object using `getattr()`. This allows you to track system resource consumption per request.
# Import the Ollama Python client.
import ollama

# Use one model consistently throughout the notebook.
OLLAMA_MODEL = "gpt-oss:20b"


def run_hello_llm(prompt_text: str) -> str:
    """
    Send a prompt to the local Ollama model and return its response.

    prompt_text:
        The question/instruction sent to the local model.

    Returns:
        The text generated by the local model.
    """

    # Send the prompt to Ollama running on the local machine.
    response = ollama.chat(
        model=OLLAMA_MODEL,
        messages=[
            {
                "role": "user",
                "content": prompt_text
            }
        ],
        options={
            # Keep generation focused and relatively deterministic.
            "temperature": 0.2,
            # Limit generation for this one-word experiment.
            "num_predict": 1000
        }
    )

    # Ollama reports local inference statistics in the response.
    # prompt_eval_count = input tokens evaluated locally.
    # eval_count = output tokens generated locally.
    prompt_tokens = getattr(response, "prompt_eval_count", 0)
    completion_tokens = getattr(response, "eval_count", 0)

    # Calculate total locally processed tokens for visibility.
    total_tokens = prompt_tokens + completion_tokens

    print(
        f"Prompt tokens: {prompt_tokens} | "
        f"Completion tokens: {completion_tokens} | "
        f"Total tokens: {total_tokens}"
    )

    # Extract and return only the generated text.
    return response.message.content


# Keep the prompt short to demonstrate prompt-efficiency principles.
prompt = "What is the core benefit of RAG? Answer with exactly one word."

# Run the local model.
result = run_hello_llm(prompt)

# Display the generated response.
print("\nLLM Response:")
print(result)
````

## 3. Week 2: Typed Contracts (Pydantic) & Async Concurrency Pipeline
* **Slide Source:** *Week 2 - Slide 05-07 ("Pydantic Models"), Slide 09-11 ("Async Basics & httpx"), Slide 15-18 ("Concurrency Patterns: Gather, Batching, Retry"), Slide 23 ("Structured Logging"), Slide 27 ("SQLite Store")*
* **Action:** Build Pydantic schemas, an async client with exponential retry backoff, parallel batch processing via `asyncio.gather`, JSON structured logging, and SQLite persistence.

````python
# Slide Source: Week 2 - Slide 05-07 ("Pydantic Models"), Slide 09-11 ("Async Basics & httpx"),
# Slide 15-18 ("Concurrency Patterns: Gather, Batching, Retry"), Slide 23 ("Structured Logging"), Slide 27 ("SQLite Store")
#
# Action: Build Pydantic schemas, an async Ollama client with exponential retry
# backoff, parallel batch processing via asyncio.gather, JSON structured logging,
# and SQLite persistence.
#
# Ollama replaces the paid cloud API for this notebook. The model runs locally.
# Frequently Asked Interview Questions:
# Q: How do you handle database concurrency and avoid "database locked" errors in SQLite when processing asynchronous LLM batches?
# A: You enable Write-Ahead Logging (`PRAGMA journal_mode=WAL;`), utilize a global `asyncio.Lock()`, and offload the blocking synchronous SQLite write operation to a separate worker thread using `asyncio.to_thread(_sync_save_to_db, record)`.
import asyncio
import json
import logging
import sqlite3
import time
import os
from typing import List
from ollama import AsyncClient
from pydantic import BaseModel, Field

# Use the same local model throughout the notebook.
OLLAMA_MODEL = "gpt-oss:20b"

# Global lock to serialize database writes across concurrent tasks
db_lock = asyncio.Lock()

# Structured Logging Setup
logging.basicConfig(level=logging.INFO, format="%(message)s")


def log_json(event: str, **kwargs):
    log_entry = {"event": event, "timestamp": time.time(), **kwargs}
    logging.info(json.dumps(log_entry))


# Pydantic Schemas
class QuestionRequest(BaseModel):
    id: int
    query: str = Field(..., min_length=3, description="The user query")


class AnswerResponse(BaseModel):
    id: int
    query: str
    answer: str
    status: str = "success"


# --- DB Initialization & Unlock Logic ---

def unlock_and_init_db(db_name="results.db"):
    """
    Checks for locks and initializes the DB.
    If heavily locked by a stale Jupyter thread, it forces an unlock.
    """
    max_retries = 3
    
    for attempt in range(1, max_retries + 1):
        try:
            # Short timeout to quickly check for stale locks
            conn = sqlite3.connect(db_name, timeout=5.0)
            cursor = conn.cursor()

            # Enable Write-Ahead Logging (WAL) mode for drastically better concurrency
            cursor.execute("PRAGMA journal_mode=WAL;")

            cursor.execute(
                """
                CREATE TABLE IF NOT EXISTS run_results (
                    id INTEGER PRIMARY KEY,
                    query TEXT NOT NULL,
                    answer TEXT NOT NULL,
                    created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
                )
                """
            )
            conn.commit()
            conn.close()
            print(f"Database '{db_name}' initialized and verified unlocked.")
            return  # Success, exit function
            
        except sqlite3.OperationalError as e:
            if "locked" in str(e).lower():
                print(f"Database locked (Attempt {attempt}/{max_retries}). Retrying in 2 seconds...")
                time.sleep(2)
            else:
                raise
                
    # If we exhaust retries, forcefully unlock by clearing stale WAL/DB files 
    # (Safe for local notebook experimentation; use caution in production)
    print("Database remained locked. Forcefully clearing lock for notebook environment...")
    for ext in ["", "-wal", "-shm"]:
        filepath = f"{db_name}{ext}"
        if os.path.exists(filepath):
            try:
                os.remove(filepath)
                print(f"Removed stale file: {filepath}")
            except OSError as cleanup_err:
                print(f"Failed to remove {filepath}: {cleanup_err}")
                
    # Re-run initialization now that files are cleared
    print("Re-initializing fresh database...")
    unlock_and_init_db(db_name)


def _sync_save_to_db(record: AnswerResponse):
    """Synchronous SQLite write operation with extended timeout and guaranteed closure."""
    conn = sqlite3.connect("results.db", timeout=20.0)
    
    try:
        cursor = conn.cursor()

        cursor.execute(
            "INSERT INTO run_results (id, query, answer) VALUES (?, ?, ?)",
            (record.id, record.query, record.answer),
        )

        conn.commit()
    finally:
        # Ensure the lock is released when the write is complete
        conn.close()


async def save_to_db(record: AnswerResponse):
    """Thread-safe, non-blocking async wrapper around DB writes."""
    async with db_lock:
        # Offload blocking SQLite I/O to a worker thread
        await asyncio.to_thread(_sync_save_to_db, record)


# Retry Wrapper with Exponential Backoff
async def with_retry(coro_func, *args, max_retries: int = 3, **kwargs):
    for attempt in range(1, max_retries + 1):
        try:
            return await coro_func(*args, **kwargs)
        except Exception as exc:
            log_json(
                "async_retry_attempt",
                attempt=attempt,
                max_retries=max_retries,
                error=str(exc),
            )

            if attempt == max_retries:
                raise

            await asyncio.sleep(2**attempt)


async def process_single_query(
    async_client: AsyncClient, req: QuestionRequest
) -> AnswerResponse:
    async def _call():
        response = await async_client.chat(
            model=OLLAMA_MODEL,
            messages=[{"role": "user", "content": req.query}],
            options={"temperature": 0.3},
        )

        # Extract message content safely
        if hasattr(response, "message"):
            return response.message.content.strip()
        return response["message"]["content"].strip()

    answer_text = await with_retry(_call)

    res = AnswerResponse(id=req.id, query=req.query, answer=answer_text)

    # Safely save to DB asynchronously without lock contention
    await save_to_db(res)

    log_json("query_processed", id=req.id, query=req.query)
    return res


async def batch_process_pipeline(queries: List[QuestionRequest]):
    async_client = AsyncClient()

    tasks = [process_single_query(async_client, q) for q in queries]

    results = await asyncio.gather(*tasks, return_exceptions=True)
    return results


# Driver Code Execution
async def main():
    sample_queries = [
        QuestionRequest(id=1, query="What is Pydantic in Python?"),
        QuestionRequest(
            id=2, query="How does asyncio.gather enable concurrent local model calls?"
        ),
        QuestionRequest(
            id=3, query="Why use SQLite for local execution persistence?"
        ),
    ]

    print("Starting Async Local Ollama Batch Pipeline...")
    batch_results = await batch_process_pipeline(sample_queries)

    for res in batch_results:
        if isinstance(res, Exception):
            print(f"\nBatch item failed: {res}")
        else:
            print(
                f"\nID: {res.id}"
                f"\nQuery: {res.query}"
                f"\nAnswer: {res.answer[:100]}..."
            )


if __name__ == "__main__":
    # 1. Unlock and initialize the DB *before* starting the main async loop
    unlock_and_init_db()
    
    # 2. Proceed into main functionality
    if "get_ipython" in globals():
        import nest_asyncio
        nest_asyncio.apply()
        asyncio.run(main())
    else:
        asyncio.run(main())
````

````python
%%writefile main.py
# Slide Source: Week 3 - Slide 06 ("FastAPI 12-line App"), Slide 08 ("/health Endpoint"), Slide 12-14 ("Streaming via StreamingResponse")
#
# Action: Build FastAPI backend supporting a health probe, a structured POST
# endpoint, and a streaming response backed by a local Ollama model.
#
# Run locally with:
#   uvicorn main:app --reload --port 8000
# Frequently Asked Interview Questions:
# Q: How do you implement real-time token streaming for an LLM endpoint in FastAPI?
# A: You create an asynchronous generator (`stream_generator`) that requests a stream from the LLM client (`stream=True`) and yields text chunks as they arrive. This generator is then passed to FastAPI's `StreamingResponse` with a `text/plain` media type.
from fastapi import FastAPI, HTTPException
from fastapi.responses import StreamingResponse
from pydantic import BaseModel
from ollama import AsyncClient

# The local Ollama model used by this backend.
OLLAMA_MODEL = "gpt-oss:20b"

# Create the FastAPI application.
app = FastAPI(
    title="Agentic RAG Engine API",
    version="2.0.0"
)

# Create one asynchronous Ollama client.
# Ollama communicates with the local Ollama service.
async_client = AsyncClient()


class AskRequest(BaseModel):
    # Text supplied by the API caller.
    query: str


@app.get("/health")
async def health_check():
    # Lightweight application health probe.
    # This does not invoke the LLM.
    return {
        "status": "ok",
        "service": "agentic-rag-engine",
        "model": OLLAMA_MODEL
    }


@app.post("/ask")
async def ask_endpoint(request: AskRequest):
    # Reject empty or whitespace-only queries.
    if not request.query.strip():
        raise HTTPException(
            status_code=400,
            detail="Query string cannot be empty."
        )

    # Send the request to the local Ollama model.
    response = await async_client.chat(
        model=OLLAMA_MODEL,
        messages=[
            {"role": "user", "content": request.query}
        ]
    )

    # Return the original query and generated answer as JSON.
    return {
        "query": request.query,
        "answer": response.message.content
    }


async def stream_generator(query: str):
    # Request streaming generation from the local Ollama model.
    response = await async_client.chat(
        model=OLLAMA_MODEL,
        messages=[
            {"role": "user", "content": query}
        ],
        stream=True
    )

    # Yield generated text chunks as they arrive.
    async for chunk in response:
        content = chunk.message.content
        if content:
            yield content


@app.post("/stream")
async def stream_endpoint(request: AskRequest):
    # Return a streaming HTTP response to the caller.
    return StreamingResponse(
        stream_generator(request.query),
        media_type="text/plain"
    )
````

### Building a Resilient, Asynchronous LLM Pipeline
An architectural breakdown of the Week 2 local Ollama batch processor.

The provided code represents a robust, production-ready pattern for communicating with Local LLMs (Large Language Models) in an asynchronous environment. Instead of making slow, sequential requests, this pipeline processes multiple prompts concurrently while enforcing strict data types and handling failures gracefully.

Here is a breakdown of how the different components work together:

Enforcing Strict Contracts with Pydantic
In agentic workflows, unpredictable data structures lead to broken pipelines. The code uses Pydantic BaseModel classes (Question and Answer) to create strict data contracts. This ensures that every piece of data moving through the pipeline conforms to an expected format, reducing the risk of runtime errors when passing data between the UI, the LLM, and the database.

The Async LLM Engine
The async_ask_ollama function is the core engine. It swaps out the original lab's fake sleep timers for genuine, asynchronous HTTP calls to a local Ollama model (gpt-oss:20b) using ollama.AsyncClient().

To effectively test the resilience of the system, this function also features a deliberate fail_rate parameter. By artificially triggering TransientError exceptions, developers can simulate real-world network hiccups, API rate limits, or locked databases without needing to actually break their local server.

Resilience via Exponential Backoff
AI generation can be resource-intensive, and requests occasionally time out or fail. The ask_llm_with_retry wrapper prevents a single failed request from crashing the entire batch process.

It catches simulated transient failures.

It pauses before trying again using an exponential backoff strategy (waiting 1 second, then 2 seconds, then 4 seconds).

This prevents overwhelming an already struggling server with immediate, repeated requests.

Concurrency Pattern: Batch vs. Stream
The code provides two distinct ways to handle concurrent LLM requests, both utilizing Python's asyncio library:

Strict Ordering (run_batch): Uses asyncio.gather to fire off all questions simultaneously, but waits for the absolute slowest response to finish before returning anything. The output strictly matches the order of the input.

Speed to Insight (run_batch_stream): Uses asyncio.as_completed. As soon as any single LLM call finishes—regardless of its position in the original list—it is immediately processed and printed. This is crucial for user-facing applications (like Streamlit UIs) where you want to show progress immediately rather than making the user stare at a loading screen until the entire batch is done.

````python
# Week 2 - Typed Contracts, Input/Output Pydantic Validation Engine & Async Pipeline
# Frequently Asked Interview Questions:
# Q: How can Pydantic be used to enforce anti-hallucination guardrails on an LLM's output?
# A: You define a `BaseModel` (like `LLMResponseSchema`) representing the target output. During runtime parsing, you compare the LLM's cited references (e.g., `referenced_txn_ids`) against a known, pre-validated set of source IDs. If a hallucinated ID is cited, you raise a `ValueError` to reject the response.

import asyncio
import random
import json
import re
import os
import traceback
from datetime import datetime
from collections import defaultdict
from pydantic import BaseModel, ConfigDict, Field, field_validator, ValidationError
from typing import Optional, List, Callable, Any
from ollama import AsyncClient


# ─── 1. RAG Input Data Validation Engine ────────────────────────────────

class TransactionRecord(BaseModel):
    """
    1) Objective: Enforce strict structural rules on incoming raw RAG data.
    2) What the code does: Uses Pydantic to validate data types, prevent negative amounts, and catch corrupted timestamps. Rejects unknown keys.
    3) Data dependency: Receives raw dictionary records parsed from a JSON string.
    4) Where it is called: Inside the `process_rag_context` function during the iteration over raw records.
    5) Output dependency: Produces a validated Python object. If validation fails, it raises a ValidationError which routes the record to the REJECTED bucket.
    """
    model_config = ConfigDict(extra='forbid') 
    
    transaction_id: str
    customer_id: str
    amount: float
    currency: str
    store_location: Optional[str] = None
    timestamp: str
    payment_method: Optional[str] = None
    status: str

    @field_validator('amount')
    @classmethod
    def validate_amount(cls, value: float) -> float:
        if value < 0:
            raise ValueError(f"Negative amount detected: {value}")
        return value

    @field_validator('timestamp')
    @classmethod
    def validate_timestamp(cls, value: str) -> str:
        if value == "Bad-Time":
            raise ValueError("Corrupted timestamp format ('Bad-Time')")
        return value


def process_rag_context(raw_json_str: str) -> tuple[str, set[str]]:
    """
    1) Objective: Sanitize and categorize input data before feeding it to the LLM.
    2) What the code does: Parses a JSON string, passes each record through `TransactionRecord`, and buckets them into ACCEPTED, REJECTED, or IGNORED. Extracts valid IDs for anti-hallucination checks.
    3) Data dependency: Takes a raw JSON string loaded directly from the physical file.
    4) Where it is called: In the Execution Block (Step 1) before `main()` begins.
    5) Output dependency: Returns a sanitized JSON string of categorized data and a Set of valid IDs. These are injected into the `Question` objects passed to the LLM.
    """
    try:
        raw_records = json.loads(raw_json_str)
    except json.JSONDecodeError as e:
        print(f"⚠️ [Input Data Error] Failed to parse JSON: {e}")
        return "{}", set()

    categorized_data = {"ACCEPTED": [], "REJECTED": [], "IGNORED": []}
    valid_txn_ids = set()

    for record in raw_records:
        try:
            valid_txn = TransactionRecord(**record)
            if valid_txn.status.lower() == "unsuccessful":
                categorized_data["IGNORED"].append(valid_txn.model_dump())
            else:
                categorized_data["ACCEPTED"].append(valid_txn.model_dump())
                valid_txn_ids.add(valid_txn.transaction_id)
        except ValidationError as e:
            categorized_data["REJECTED"].append({
                "original_record": record,
                "validation_errors": [err["msg"] for err in e.errors()]
            })
            
    return json.dumps(categorized_data, indent=2), valid_txn_ids


# ─── 2. Response Validation Contract (LLM Output Guardrail) ──────────────

class LLMResponseSchema(BaseModel):
    """
    1) Objective: Define the exact target structure expected back from the LLM.
    2) What the code does: Establishes a schema with a summary, list of cited IDs, and a confidence score. Prevents empty string answers.
    3) Data dependency: Receives a dictionary parsed from the LLM's raw output string.
    4) Where it is called: Inside `validate_and_parse_llm_response`.
    5) Output dependency: Outputs a typed Python object used to construct the final `Answer` object, or raises a ValidationError if the LLM hallucinated the structure.
    """
    answer_summary: str = Field(description="Direct, non-empty answer to the query")
    referenced_txn_ids: List[str] = Field(default_factory=list, description="IDs cited in response")
    confidence_score: float = Field(ge=0.0, le=1.0, description="Confidence rating between 0 and 1")

    @field_validator('answer_summary')
    @classmethod
    def check_non_empty(cls, value: str) -> str:
        if not value.strip():
            raise ValueError("LLM generated an empty answer summary.")
        return value.strip()


# ─── 3. Exclusive Runtime Response Validation Engine ────────────────────

def validate_and_parse_llm_response(
    raw_response_text: str, 
    valid_ids_context: Optional[set[str]] = None
) -> LLMResponseSchema:
    """
    1) Objective: Clean, parse, and enforce anti-hallucination rules on LLM outputs.
    2) What the code does: Strips markdown, parses JSON, validates against `LLMResponseSchema`, and checks if cited IDs exist in the `valid_ids_context`.
    3) Data dependency: Takes the raw string content returned by the Ollama API, and the set of valid IDs extracted during `process_rag_context`.
    4) Where it is called: Inside the `async_ask_ollama` function right after receiving the API response.
    5) Output dependency: Returns a clean `LLMResponseSchema` object to `async_ask_ollama`, or raises an error which triggers the fallback error logging.
    """
    clean_text = raw_response_text.strip()
    json_match = re.search(r'```(?:json)?\s*(\{.*?\})\s*```', clean_text, re.DOTALL)
    if json_match:
        clean_text = json_match.group(1)

    try:
        data_dict = json.loads(clean_text)
    except json.JSONDecodeError:
        data_dict = {
            "answer_summary": raw_response_text.strip(),
            "referenced_txn_ids": [],
            "confidence_score": 0.8 
        }

    validated_response = LLMResponseSchema(**data_dict)

    if valid_ids_context:
        for txn_id in validated_response.referenced_txn_ids:
            if txn_id not in valid_ids_context:
                raise ValueError(
                    f"Hallucination Detected! Transaction ID '{txn_id}' cited by LLM does not exist in context."
                )

    return validated_response


# ─── 4. Pipeline Typed Models & LLM Integration ────────────────────────

class Question(BaseModel):
    """
    1) Objective: Standardize the input request to the LLM.
    2) What the code does: Packages the user text, stringified context data, and the valid ID set.
    3) Data dependency: Instantiated manually in the Execution Block using standard strings and sets.
    4) Where it is called: Passed as an argument to `ask_llm_with_retry` and `async_ask_ollama`.
    5) Output dependency: Used by `async_ask_ollama` to build the final prompt for the LLM.
    """
    text: str
    context: Optional[str] = None
    valid_txn_ids: Optional[set[str]] = None

class Answer(BaseModel):
    """
    1) Objective: Standardize the final output result of an LLM query.
    2) What the code does: Stores the final parsed text, confidence, and flags if an error occurred.
    3) Data dependency: Populated using data from `LLMResponseSchema` or exception traces.
    4) Where it is called: Created and returned at the very end of `async_ask_ollama` and `ask_llm_with_retry`.
    5) Output dependency: Collected into a list by `run_batch_stream` and printed to the terminal.
    """
    question_text: str
    text: str
    confidence: Optional[float] = None
    is_error: bool = False
    error_type: Optional[str] = None

class TransientError(Exception):
    """
    1) Objective: Provide a specific exception for temporary, recoverable failures.
    2) What the code does: Acts as a custom exception class.
    3) Data dependency: None.
    4) Where it is called: Raised dynamically in `async_ask_ollama` or agents based on a simulated `fail_rate`.
    5) Output dependency: Caught by `ask_llm_with_retry` and `run_agent_with_retry` to trigger exponential backoff.
    """
    pass


async def async_ask_ollama(question: Question, fail_rate: float = 0.0) -> Answer:
    """
    1) Objective: Send a prompt to the local LLM and process the response.
    2) What the code does: Formats a strict JSON-enforced system prompt, calls the Ollama AsyncClient, and executes runtime validation on the result. Simulates transient errors if fail_rate > 0.
    3) Data dependency: Requires a `Question` object. Requires Ollama to be running locally with the specified model.
    4) Where it is called: Inside `ask_llm_with_retry`.
    5) Output dependency: Returns an `Answer` object which bubbles up to the stream handler.
    """
    if random.random() < fail_rate:
        await asyncio.sleep(0.1)
        raise TransientError("Simulated transient failure")
    
    system_prompt = (
        "You are a strict data assistant. Respond to the user request ONLY using a JSON object with this exact structure:\n"
        "{\n"
        '  "answer_summary": "Your detailed answer string",\n'
        '  "referenced_txn_ids": ["TXN-xxxx"],\n'
        '  "confidence_score": 0.95\n'
        "}\n"
    )
    
    messages = [{"role": "system", "content": system_prompt}]
    if question.context:
        messages.append({
            "role": "system", 
            "content": f"Context Data: {question.context}"
        })
        
    messages.append({"role": "user", "content": question.text})

    response = await AsyncClient().chat(
        model="gpt-oss:20b", 
        messages=messages
    )
    raw_content = response['message']['content']

    try:
        parsed_output = validate_and_parse_llm_response(
            raw_response_text=raw_content, 
            valid_ids_context=question.valid_txn_ids
        )
        return Answer(
            question_text=question.text,
            text=parsed_output.answer_summary,
            confidence=parsed_output.confidence_score,
            is_error=False
        )
    except (ValidationError, ValueError) as resp_err:
        err_details = str(resp_err).replace('\n', ' ')
        return Answer(
            question_text=question.text,
            text=f"Raw LLM Output: {raw_content[:60]}...",
            is_error=True,
            error_type=f"Response Validation Failed ({err_details})"
        )


# ─── 5. Async Execution & Error Loggers ─────────────────────────────────

async def ask_llm_with_retry(question: Question, fail_rate: float = 0.0, max_attempts: int = 3) -> Answer:
    """
    1) Objective: Make the LLM integration fault-tolerant.
    2) What the code does: Attempts to call `async_ask_ollama` up to `max_attempts` times, pausing with exponential backoff if a `TransientError` is caught.
    3) Data dependency: Takes a `Question` object.
    4) Where it is called: Inside list comprehensions in `run_batch` and `run_batch_stream`.
    5) Output dependency: Returns the successful `Answer` or a fabricated error `Answer` if retries are exhausted.
    """
    for attempt in range(max_attempts):
        try:
            return await async_ask_ollama(question, fail_rate=fail_rate)
        except TransientError:
            if attempt == max_attempts - 1:
                return Answer(
                    question_text=question.text,
                    text="Max retries reached due to transient network failures.",
                    is_error=True,
                    error_type="Network Transient Fail"
                )
            backoff = 2 ** attempt  
            print(f"    [Retry] Attempt {attempt + 1} failed. Retrying in {backoff}s...")
            await asyncio.sleep(backoff)
        except Exception as e:
            error_msg = f"{type(e).__name__}: {str(e)}"
            return Answer(
                question_text=question.text, 
                text=f"Execution aborted: {error_msg}", 
                is_error=True,
                error_type="System API Exception"
            )

async def run_batch(questions: list[Question], fail_rate: float = 0.0) -> list[Answer]:
    """
    1) Objective: Process multiple questions concurrently (Wait for all to finish).
    2) What the code does: Uses `asyncio.gather` to run tasks simultaneously, maintaining original input order.
    3) Data dependency: A list of `Question` objects.
    4) Where it is called: (Available for use, but currently superseded by `run_batch_stream` in `main`).
    5) Output dependency: Returns a list of `Answer` objects.
    """
    tasks = [ask_llm_with_retry(q, fail_rate=fail_rate) for q in questions]
    return await asyncio.gather(*tasks)
    
async def run_batch_stream(questions: list[Question], fail_rate: float = 0.0) -> list[Answer]:
    """
    1) Objective: Process multiple questions concurrently and stream outputs as they finish.
    2) What the code does: Uses `asyncio.as_completed` to yield results immediately upon completion, rather than waiting for the slowest task. Prints formatting to console.
    3) Data dependency: A list of `Question` objects.
    4) Where it is called: In `main()`, scheduled as an `asyncio.create_task`.
    5) Output dependency: Returns a collected list of `Answer` objects which is awaited by `asyncio.gather` in `main()`.
    """
    tasks = [ask_llm_with_retry(q, fail_rate=fail_rate) for q in questions]
    results: list[Answer] = []

    for coro in asyncio.as_completed(tasks):
        try:
            ans = await coro
            print(f"\n🤖 [LLM Agent Task Output]")
            print(f"   ❓ Question: {ans.question_text}")
            if ans.is_error:
                print(f"   ❌ STATUS: FAILED | Type: {ans.error_type}")
                print(f"      Log Detail: {ans.text}")
            else:
                print(f"   ✅ STATUS: PASSED | Confidence: {ans.confidence}")
                print(f"      Answer:   {ans.text}") 
            print("-" * 65)  
            results.append(ans)
        except Exception as e:
            print(f"💥 [Critical Unhandled Loop Error]: {e}")
            traceback.print_exc()
        
    return results


# ─── 6. Ten Independent Agents with Search Logic & Retry Engine ──────────

async def run_agent_with_retry(
    agent_func: Callable[[list], Any], 
    data: list, 
    agent_name: str, 
    objective: str, 
    fail_rate: float = 0.0, 
    max_attempts: int = 3
) -> str:
    """
    1) Objective: Provide a universal fault-tolerant wrapper for all background agents.
    2) What the code does: Executes a specific agent function. Implements exponential backoff for `TransientError` and graceful exception handling so parallel execution never crashes.
    3) Data dependency: Requires the target agent function, the parsed JSON `data` list, and string identifiers.
    4) Where it is called: In `main()`, inside `asyncio.create_task` wrappers for all 10 agents.
    5) Output dependency: Returns the final string result of the agent, or a formatted error message string.
    """
    print(f"\n🚀 [{agent_name} Initialized]")
    print(f"   🎯 Objective: {objective}")
    
    for attempt in range(max_attempts):
        try:
            if random.random() < fail_rate:
                await asyncio.sleep(0.05)
                raise TransientError("Simulated agent transient processing glitch")
                
            result = await agent_func(data)
            print(f"   ✅ [{agent_name} Completed]")
            print(f"      Result: {result}")
            print("-" * 65)
            return result
        except TransientError as te:
            if attempt == max_attempts - 1:
                err_msg = f"Max retries ({max_attempts}) reached due to: {te}"
                print(f"   ❌ [{agent_name} Failed] {err_msg}")
                print("-" * 65)
                return f"Error: {err_msg}"
            backoff = 2 ** attempt
            print(f"   ⚠️ [{agent_name}] Attempt {attempt + 1} transient error. Retrying in {backoff}s...")
            await asyncio.sleep(backoff)
        except Exception as e:
            err_msg = f"Unhandled Exception handled gracefully -> {type(e).__name__}: {str(e)}"
            print(f"   💥 [{agent_name} Exception] {err_msg}")
            print("-" * 65)
            return err_msg


async def agent_data_quality(data: list) -> str:
    """
    1) Objective: Identify malformed data in the JSON dataset.
    2) What the code does: Filters for "Bad-Time" timestamps and None/null store locations.
    3) Data dependency: The raw list of dictionaries parsed from `retail_payment_records.json`.
    4) Where it is called: Passed as `agent_func` into `run_agent_with_retry` from `main()`.
    5) Output dependency: Returns a summary string logged to the console.
    """
    await asyncio.sleep(0.1) 
    corrupt_timestamps = [t['transaction_id'] for t in data if t.get('timestamp') == "Bad-Time"]
    missing_locations = [t['transaction_id'] for t in data if t.get('store_location') is None]
    return f"Found {len(corrupt_timestamps)} corrupt timestamps (e.g. {corrupt_timestamps}) & {len(missing_locations)} null store locations."

async def agent_fraud_detection(data: list) -> str:
    """
    1) Objective: Flag financially impossible scenarios.
    2) What the code does: Scans the dataset for transaction amounts less than zero.
    3) Data dependency: The raw list of dictionaries parsed from `retail_payment_records.json`.
    4) Where it is called: Passed as `agent_func` into `run_agent_with_retry` from `main()`.
    5) Output dependency: Returns a summary string logging anomalous Transaction IDs.
    """
    await asyncio.sleep(0.12)
    anomalies = [t for t in data if float(t.get('amount', 0)) < 0]
    flagged_ids = [a['transaction_id'] for a in anomalies]
    return f"Detected {len(anomalies)} negative amount anomalies in file! Flagged Txn IDs: {flagged_ids}"

async def agent_gateway_health(data: list) -> str:
    """
    1) Objective: Assess reliability of payment methods.
    2) What the code does: Counts occurrences where `is_successful` is False, grouped by `payment_method`.
    3) Data dependency: The raw list of dictionaries parsed from `retail_payment_records.json`.
    4) Where it is called: Passed as `agent_func` into `run_agent_with_retry` from `main()`.
    5) Output dependency: Returns a formatted dictionary string of failure counts.
    """
    await asyncio.sleep(0.08)
    failures = defaultdict(int)
    for t in data:
        if not t.get('is_successful', True):
            failures[t.get('payment_method', 'Unknown')] += 1
    return f"Failed transactions by payment gateway: {dict(failures)}"

async def agent_financial_recon(data: list) -> str:
    """
    1) Objective: Perform aggregate accounting.
    2) What the code does: Sums up all successful transaction amounts grouped by their respective currency.
    3) Data dependency: The raw list of dictionaries parsed from `retail_payment_records.json`.
    4) Where it is called: Passed as `agent_func` into `run_agent_with_retry` from `main()`.
    5) Output dependency: Returns a dictionary string showing the ledger totals.
    """
    await asyncio.sleep(0.15)
    recon = defaultdict(float)
    for t in data:
        if t.get('is_successful', True) and float(t.get('amount', 0)) > 0:
            recon[t.get('currency', 'USD')] += float(t.get('amount', 0))
    formatted_recon = {curr: round(total, 2) for curr, total in recon.items()}
    return f"Reconciled Total Revenue by Currency: {formatted_recon}"

async def agent_geo_insights(data: list) -> str:
    """
    1) Objective: Compare physical vs. digital performance.
    2) What the code does: Aggregates successful transaction totals based on `store_location`. Nulls become "Online/E-Commerce".
    3) Data dependency: The raw list of dictionaries parsed from `retail_payment_records.json`.
    4) Where it is called: Passed as `agent_func` into `run_agent_with_retry` from `main()`.
    5) Output dependency: Returns a string representation of revenue split by physical location.
    """
    await asyncio.sleep(0.1)
    location_revenue = defaultdict(float)
    for t in data:
        if t.get('is_successful', True) and float(t.get('amount', 0)) > 0:
            loc = t.get('store_location') or 'Online/E-Commerce'
            location_revenue[loc] += float(t.get('amount', 0))
    formatted_geo = {loc: round(amt, 2) for loc, amt in location_revenue.items()}
    return f"Revenue by Location: {formatted_geo}"

async def agent_loyalty(data: list) -> str:
    """
    1) Objective: Identify top customers for rewards programs.
    2) What the code does: Sums all successful spends grouped by `customer_id`, then finds the maximum value.
    3) Data dependency: The raw list of dictionaries parsed from `retail_payment_records.json`.
    4) Where it is called: Passed as `agent_func` into `run_agent_with_retry` from `main()`.
    5) Output dependency: Returns a string stating the highest spending VIP ID and total.
    """
    await asyncio.sleep(0.2)
    customer_spend = defaultdict(float)
    for t in data:
        if t.get('is_successful', True) and float(t.get('amount', 0)) > 0:
            customer_spend[t.get('customer_id')] += float(t.get('amount', 0))
    top_cust = max(customer_spend, key=customer_spend.get) if customer_spend else "None"
    return f"Top VIP Customer identified: {top_cust} with total spend of ${round(customer_spend[top_cust], 2)}"

async def agent_abandoned_cart(data: list) -> str:
    """
    1) Objective: Prep data for re-engagement marketing.
    2) What the code does: Filters for unsuccessful transactions and extracts a unique set of `customer_id`s.
    3) Data dependency: The raw list of dictionaries parsed from `retail_payment_records.json`.
    4) Where it is called: Passed as `agent_func` into `run_agent_with_retry` from `main()`.
    5) Output dependency: Returns a summary string showing the count of queued customers.
    """
    await asyncio.sleep(0.09)
    failed_customers = set(t['customer_id'] for t in data if not t.get('is_successful', True))
    return f"Identified {len(failed_customers)} unique customers with failed payments queued for re-engagement."

async def agent_currency_conversion(data: list) -> str:
    """
    1) Objective: Create a single global revenue metric.
    2) What the code does: Normalizes all successful transactions into a single USD value using a static exchange rate map.
    3) Data dependency: The raw list of dictionaries parsed from `retail_payment_records.json`.
    4) Where it is called: Passed as `agent_func` into `run_agent_with_retry` from `main()`.
    5) Output dependency: Returns a string declaring the combined global revenue in USD.
    """
    await asyncio.sleep(0.22)
    rates = {"USD": 1.0, "EUR": 1.1, "GBP": 1.25, "INR": 0.012, "YEN": 0.0067}
    total_usd = 0.0
    for t in data:
        if t.get('is_successful', True) and float(t.get('amount', 0)) > 0:
            curr = t.get('currency', 'USD')
            total_usd += float(t.get('amount', 0)) * rates.get(curr, 1.0)
    return f"Normalized Global Gross Revenue: ${round(total_usd, 2)} USD"

async def agent_temporal_analysis(data: list) -> str:
    """
    1) Objective: Discover peak trading periods.
    2) What the code does: Parses ISO strings to datetimes, extracts the day of the week, and finds the day with the highest transaction volume.
    3) Data dependency: The raw list of dictionaries parsed from `retail_payment_records.json`.
    4) Where it is called: Passed as `agent_func` into `run_agent_with_retry` from `main()`.
    5) Output dependency: Returns a string indicating the busiest day.
    """
    await asyncio.sleep(0.14)
    weekday_counts = defaultdict(int)
    for t in data:
        ts = t.get('timestamp')
        if ts and ts != "Bad-Time":
            try:
                dt = datetime.fromisoformat(ts.replace("Z", "+00:00"))
                weekday_counts[dt.strftime("%A")] += 1
            except ValueError:
                pass
    peak_day = max(weekday_counts, key=weekday_counts.get) if weekday_counts else "Unknown"
    return f"Busiest sales day identified: {peak_day} ({weekday_counts[peak_day]} transactions recorded)"

async def agent_compliance_audit(data: list) -> str:
    """
    1) Objective: Ensure Anti-Money Laundering (AML) regulatory compliance.
    2) What the code does: Filters specifically for Cash transactions greater than a $1000 threshold.
    3) Data dependency: The raw list of dictionaries parsed from `retail_payment_records.json`.
    4) Where it is called: Passed as `agent_func` into `run_agent_with_retry` from `main()`.
    5) Output dependency: Returns a string with the IDs requiring manual auditing.
    """
    await asyncio.sleep(0.11)
    aml_flags = [
        t['transaction_id'] for t in data 
        if t.get('payment_method') == "Cash" and float(t.get('amount', 0)) > 1000
    ]
    return f"Flagged {len(aml_flags)} high-value cash transactions for AML compliance audit (IDs: {aml_flags})"


# ─── 7. Execution Block & File Handling ─────────────────────────────────

RAW_150_RECORDS = [
    {"transaction_id": "TXN-1001", "customer_id": "CUST-524", "amount": 246.77, "currency": "USD", "payment_method": "UPI", "timestamp": "2026-08-13T05:43:49.141240", "is_successful": True, "store_location": "New York"},
    {"transaction_id": "TXN-1002", "customer_id": "CUST-858", "amount": 945.15, "currency": "EUR", "payment_method": "UPI", "timestamp": "2026-08-04T05:43:49.141240", "is_successful": True, "store_location": "Mumbai"},
    {"transaction_id": "TXN-1003", "customer_id": "CUST-316", "amount": 657.43, "currency": "INR", "payment_method": "Debit Card", "timestamp": "2026-08-02T05:43:49.141240", "is_successful": True, "store_location": "Mumbai"},
    {"transaction_id": "TXN-1004", "customer_id": "CUST-354", "amount": 1051.77, "currency": "USD", "payment_method": "UPI", "timestamp": "2026-08-03T05:43:49.141240", "is_successful": True, "store_location": "London"},
    {"transaction_id": "TXN-1005", "customer_id": "CUST-240", "amount": 954.45, "currency": "INR", "payment_method": "PayPal", "timestamp": "2026-08-04T05:43:49.141240", "is_successful": True, "store_location": None},
    {"transaction_id": "TXN-1006", "customer_id": "CUST-968", "amount": 929.12, "currency": "EUR", "payment_method": "Debit Card", "timestamp": "2026-08-15T05:43:49.141240", "is_successful": True, "store_location": None},
    {"transaction_id": "TXN-1007", "customer_id": "CUST-453", "amount": 1168.83, "currency": "INR", "payment_method": "UPI", "timestamp": "2026-08-21T05:43:49.141240", "is_successful": True, "store_location": None},
    {"transaction_id": "TXN-1008", "customer_id": "CUST-750", "amount": 179.92, "currency": "EUR", "payment_method": "PayPal", "timestamp": "2026-08-14T05:43:49.141240", "is_successful": True, "store_location": "Mumbai"},
    {"transaction_id": "TXN-1009", "customer_id": "CUST-921", "amount": 1154.48, "currency": "INR", "payment_method": "Debit Card", "timestamp": "2026-08-01T05:43:49.141240", "is_successful": True, "store_location": "New York"},
    {"transaction_id": "TXN-1010", "customer_id": "CUST-631", "amount": 1320.56, "currency": "INR", "payment_method": "UPI", "timestamp": "2026-07-27T05:43:49.141240", "is_successful": True, "store_location": "New York"},
    {"transaction_id": "TXN-1011", "customer_id": "CUST-679", "amount": -25.5, "currency": "EUR", "payment_method": "Credit Card", "timestamp": "2026-08-03T05:43:49.141240", "is_successful": True, "store_location": "London"},
    {"transaction_id": "TXN-1012", "customer_id": "CUST-693", "amount": 798.49, "currency": "USD", "payment_method": "PayPal", "timestamp": "2026-08-03T05:43:49.141240", "is_successful": True, "store_location": "Berlin"},
    {"transaction_id": "TXN-1013", "customer_id": "CUST-374", "amount": 795.13, "currency": "EUR", "payment_method": "Debit Card", "timestamp": "2026-08-15T05:43:49.141240", "is_successful": True, "store_location": "New York"},
    {"transaction_id": "TXN-1014", "customer_id": "CUST-618", "amount": 878.46, "currency": "EUR", "payment_method": "PayPal", "timestamp": "2026-08-10T05:43:49.141240", "is_successful": True, "store_location": "London"},
    {"transaction_id": "TXN-1015", "customer_id": "CUST-684", "amount": 47.53, "currency": "USD", "payment_method": "PayPal", "timestamp": "2026-08-21T05:43:49.141240", "is_successful": True, "store_location": None},
    {"transaction_id": "TXN-1016", "customer_id": "CUST-780", "amount": 851.32, "currency": "USD", "payment_method": "Debit Card", "timestamp": "2026-08-07T05:43:49.141240", "is_successful": False, "store_location": "New York"},
    {"transaction_id": "TXN-1017", "customer_id": "CUST-488", "amount": 212.47, "currency": "EUR", "payment_method": "PayPal", "timestamp": "2026-08-18T05:43:49.141240", "is_successful": True, "store_location": "Berlin"},
    {"transaction_id": "TXN-1018", "customer_id": "CUST-217", "amount": 105.93, "currency": "INR", "payment_method": "Cash", "timestamp": "2026-08-03T05:43:49.141240", "is_successful": True, "store_location": "Berlin"},
    {"transaction_id": "TXN-1019", "customer_id": "CUST-674", "amount": 1465.93, "currency": "EUR", "payment_method": "Credit Card", "timestamp": "2026-08-04T05:43:49.141240", "is_successful": True, "store_location": None},
    {"transaction_id": "TXN-1020", "customer_id": "CUST-180", "amount": 1494.87, "currency": "GBP", "payment_method": "Cash", "timestamp": "2026-08-24T05:43:49.141240", "is_successful": True, "store_location": "Berlin"},
    {"transaction_id": "TXN-1021", "customer_id": "CUST-621", "amount": 1204.22, "currency": "INR", "payment_method": "PayPal", "timestamp": "2026-08-08T05:43:49.141240", "is_successful": False, "store_location": "Mumbai"},
    {"transaction_id": "TXN-1028", "customer_id": "CUST-176", "amount": 47.25, "currency": "INR", "payment_method": "UPI", "timestamp": "2026-08-23T05:43:49.141240", "is_successful": False, "store_location": "Mumbai"},
    {"transaction_id": "TXN-1101", "customer_id": "CUST-645", "amount": 762.8, "currency": "EUR", "payment_method": "PayPal", "timestamp": "Bad-Time", "is_successful": False, "store_location": "London"}
]

def ensure_json_file_exists(filename: str = "retail_payment_records.json"):
    """
    1) Objective: Ensure a clean, working environment for script execution.
    2) What the code does: Checks if the target JSON file exists; if not, constructs it using `RAW_150_RECORDS`.
    3) Data dependency: Hardcoded Python list `RAW_150_RECORDS`.
    4) Where it is called: Very beginning of the Execution Block.
    5) Output dependency: Creates a physical `.json` file that is immediately read back by the system.
    """
    if not os.path.exists(filename):
        with open(filename, "w", encoding="utf-8") as f:
            json.dump(RAW_150_RECORDS, f, indent=4)
        print(f"📁 Created '{filename}' at local path with initial transaction records.")


# Execute Setup: Ensure file exists before reading it.
file_path = "retail_payment_records.json"
ensure_json_file_exists(file_path)

# Execute Data Load: Read the JSON file into a string for the LLM pipeline, and dictionary list for the Agents.
try:
    with open(file_path, "r", encoding="utf-8") as f:
        raw_json_string = f.read()
        file_data = json.loads(raw_json_string)
    print(f"📖 Loaded {len(file_data)} records directly from '{file_path}'\n" + "=" * 70)
except Exception as e:
    print(f"❌ Failed to load JSON file '{file_path}': {e}")
    raw_json_string = "[]"
    file_data = []

# Execute Request/Input Validation Engine (LLM Pipeline Context Prep)
filtered_rag_data, valid_txn_ids = process_rag_context(raw_json_string)

sample_questions = [
    Question(text=t, context=filtered_rag_data, valid_txn_ids=valid_txn_ids) 
    for t in [
        "What are the details of transaction TXN-1001?",
        "Which customer made a transaction with a negative amount?"
    ]
]


async def main():
    """
    1) Objective: Orchestrate the entire asynchronous pipeline.
    2) What the code does: Schedules the LLM batch stream, schedules the 10 data agents, and awaits them all concurrently.
    3) Data dependency: Pulls `file_data` and `sample_questions` from the global scope variables established directly above.
    4) Where it is called: Bottom of the script.
    5) Output dependency: Prints final aggregated LLM results and agent results to the console.
    """
    print("\nStarting Async Pipeline (LLM Stream + 10 Parallel File Search Agents)\n")
    print("=" * 70)

    # 1. First Task: run_batch_stream created as asyncio.create_task on top
    llm_stream_task = asyncio.create_task(run_batch_stream(sample_questions, fail_rate=0.0))

    # 2. Remaining 10 Agent Tasks created sequentially as create_task below it
    agent_1_task = asyncio.create_task(
        run_agent_with_retry(agent_data_quality, file_data, "Agent 1 - Data Quality", "Scan file for invalid timestamps & null values", fail_rate=0.1)
    )
    agent_2_task = asyncio.create_task(
        run_agent_with_retry(agent_fraud_detection, file_data, "Agent 2 - Fraud & Anomaly", "Search file for negative transaction amounts", fail_rate=0.1)
    )
    agent_3_task = asyncio.create_task(
        run_agent_with_retry(agent_gateway_health, file_data, "Agent 3 - Gateway Health", "Analyze failed transaction distribution by payment provider")
    )
    agent_4_task = asyncio.create_task(
        run_agent_with_retry(agent_financial_recon, file_data, "Agent 4 - Financial Recon", "Reconcile total revenue by currency")
    )
    agent_5_task = asyncio.create_task(
        run_agent_with_retry(agent_geo_insights, file_data, "Agent 5 - Geo Insights", "Analyze store location revenue vs online sales")
    )
    agent_6_task = asyncio.create_task(
        run_agent_with_retry(agent_loyalty, file_data, "Agent 6 - Loyalty & Rewards", "Identify top spending VIP customer")
    )
    agent_7_task = asyncio.create_task(
        run_agent_with_retry(agent_abandoned_cart, file_data, "Agent 7 - Abandoned Cart", "Extract customer IDs with failed payments for email recovery")
    )
    agent_8_task = asyncio.create_task(
        run_agent_with_retry(agent_currency_conversion, file_data, "Agent 8 - Currency Standardization", "Normalize global revenue into USD base currency")
    )
    agent_9_task = asyncio.create_task(
        run_agent_with_retry(agent_temporal_analysis, file_data, "Agent 9 - Temporal Analysis", "Parse ISO timestamps to determine peak transaction day")
    )
    agent_10_task = asyncio.create_task(
        run_agent_with_retry(agent_compliance_audit, file_data, "Agent 10 - Compliance Audit", "Flag cash transactions over $1000 for AML audit")
    )

    # 3. Gather all tasks at the end with await
    results = await asyncio.gather(
        llm_stream_task,
        agent_1_task,
        agent_2_task,
        agent_3_task,
        agent_4_task,
        agent_5_task,
        agent_6_task,
        agent_7_task,
        agent_8_task,
        agent_9_task,
        agent_10_task
    )

    llm_answers = results[0]
    passed = sum(1 for a in llm_answers if not a.is_error)
    failed = sum(1 for a in llm_answers if a.is_error)
    
    print("\n" + "=" * 70)
    print(f"🏁 Execution Finished. Passed LLM Validation: {passed} | Failed LLM Validation: {failed}")
    print("=" * 70)


# In Jupyter Notebook:
await main()

# In standard Python script (Commented out for Jupyter environment):
# if __name__ == "__main__":
#     asyncio.run(main())
````
